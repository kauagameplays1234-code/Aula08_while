```python
# aprendendo commo funciona while
#exemplo 01

loop = 0

while loop <= 10 :  #oque vai limitar o while
  print("loop: ", loop)
  loop += 1 #incrementar/adicionar +1 até chegar no número do while
```

```python
#exemplo 02

contador = 10

while contador > 0:
  print("contagem: ",contador)
  contador -= 1 #decrementar/tirar -1 até o número dentro do while
```

```python
#exemplo 03

soma = 0
numero = 0

while numero != -1:
  numero = int(input("digite um número: "))
  if numero != -1:
    soma += numero
print("a soma dos números: ", soma)
```

```python
#exemplo 04

senha_correta = "1234"
tentativa = ""

while tentativa != senha_correta:
  tentativa = input("digite a senha: ")
print("Acesso permitido")
```

```python
#Atividade 01 (contagem regressiva)

numero = 10

while numero > 0:
  print("contagem regressiva: ",numero)
  numero -= 1
print("fim da contagem")
```

```python
#Atividade 02 (Apenas números pares)

contador = 0

while contador <= 20:
  if contador % 2 == 0:
   print(contador)
  contador += 1
```

```python
#Atividade 03

numero = 0
soma = 0

while numero != -5:
  numero = int(input("digite um numero: "))
  if numero != -5:
   soma += numero
print("resultado da soma: ", soma)
```

```python
#Atividade 04

codigo_correto = "cofre2025"
codigo =""

while codigo != codigo_correto:
  codigo = input("digite o codigo: ")
print("Cofre aberto com sucesso")
```
